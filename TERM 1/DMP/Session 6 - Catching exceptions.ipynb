```python
try:
    fin = open('bad_file')
except FileNotFoundError:
    print('check if this file in the folder?')
except:
    print('some error')
```

```python
# when opening files, several other errors are common, for e.g. you do not have permission to access to the file
# there is no file in the current working directory with this name.
# instead of letting Python handle the exception and display a message, we are intercepting it and displaying
# this can be useful to display easier messages for the user
```

```python
try:    
    fin = open('bad_file')
    for line in fin:
        print(line)
    fin.close()
except:
    print('Something went wrong.')
```
