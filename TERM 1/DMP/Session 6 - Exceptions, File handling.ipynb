### format operator

%g to format a float number

%s to format a string

```python
temp = 24.1
# format as a string
print('%d' % temp)
```

```python
# embed into a msg
num = 2.341
print("The temperature now is %d degrees." % num) # prints as a string
```

```python
# more formats
days = 7
temp = 24.21
units = 'degrees'
print('In the recent %d days we have seen max of %g %s.' % (days, temp,units))
```

Why do we need to Catch Exceptions?

```python
fin = open('bad_file', 'r')

# when opening files, several other errors are common, for e.g. you do not have permission to access to the file
# there is no file in the current working directory with this name.
# instead of letting Python handle the exception and display a message, we are intercepting it and displaying
# this can be useful to display easier messages for the user
```

```python
try:    
    fin = open('bad_file')
    for line in fin:
        print(line)
    fin.close()
except FileNotFoundError:
    print('I know this will happen')
except:
    print('Something went wrong.')
```

```python
# how to catch a specific exception and make meaning hints available to user

lst = [1,2,3,4,5]
print(lst[6])
```

```python
try:    
    lst = [1,2,3,4,5]
    print(lst[6])
except IndexError:
    print('looks like index was outside the length of list')
except:
    print('Something went wrong.')
```

```python
# Exercise
# open a file for writing, using file1 as name
# write a line "#1" using file1
# try opening the same file for appending, using file2 as name
# append a line "#2" using file2
# close both file objects
# read the file


file1 = open("sample.txt", "w")

file1.write("#1\n")

file2 = open("sample.txt", "a")

file2.write("#2\n")

file1.close()
file2.close()

file = open("sample.txt", "r")

content = file.read()
print(content)

file.close()
```

```python
# Exercise
# five of your friends are celebrating their birthdays
# create a list of their names, use a for loop and generate personalized greetings
# write the greetings to a file
# open the file and display the greetings
```

```python
friends = ["Chirag", "Gopi", "Abdul", "Hemanth", "Subhadeep"]

file = open("birthday_greetings.txt", "w")
for friend in friends:
    greeting = f"Happy Birthday, {friend}! Have a wonderful day!\n"
    file.write(greeting)
file.close()

file = open("birthday_greetings.txt", "r")

content = file.read()
print(content)

file.close()
```

```python
# consider " " as the indicator of a new word
file = open("what_can_python_do.txt", "r")

content = file.read()

words = content.split(" ")

print("Number of words:", len(words))

file.close()
```

```python
# consider " " or ? or . as indicators of new words

file = open("what_can_python_do.txt", "r")
content = file.read()

content = content.replace("?", " ")
content = content.replace(".", " ")

words = content.split()

print("Number of words:", len(words))

file.close()
```
