```python
### Setting Seed before generating random numbers helps get the same numbers, useful to replicate results and test.

# initiate seed as 1
np.random.seed(1)

np.random.rand(2)

np.random.rand(2)
```

```python
# reset seed as 1 again
np.random.seed(1)

# reset seed as 1 again, the following command again produces same first two numbers
np.random.rand(2)

np.random.rand(2)
```

```python
# reset seed as 1 again
np.random.seed(1)

# reset seed as 1 again, the following command again produces same first two numbers
np.random.rand(2)
```

```python
# reset seed

np.random.seed(None)
np.random.rand(2)
```

```python
# np.random.normal(mean, stdev, sample size)
# this returns a sample of specifie size randomly drawn from a normal distribution with given mean and stdev
# set the seed to a specific value so every run generates the same random numbers
np.random.seed(0)

height = np.round(np.random.normal(155, 10, 1), 0)

height

np.mean(height),np.var(height),np.std(height)
```
