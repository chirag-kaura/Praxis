# using packages - pandas

The pandas name itself is derived from panel data, an econometrics term for multidimensional
structured datasets, and a play on the phrase Python data analysis.

Pandas is used for creating and manipulating easy to use data structures, it also includes data analysis tools

The pandas library essentially has three data structures:

Series:
For one-dimensional array, of any type: integers, floats, strings, and Python

DataFrame
For 2D data structure with each column of any datatypes. 
DataFrame can be created from NumPy array, Lists, Dicts, pd.Series, A 2D NumPy array

Panel
A Panel is a data structure that handles 3D data

Data Frame: is a special data object, comparable to a spreadsheet
Data Frame is made up of rows and columns. Columns can be of different types. Within a column all rows are of same type.
Hence, a data frame can be visualized as a set of arrays.

create data frame reading data from files or using pandas Dataframe() on a set of arrays. 
Matrix is a 2D structure like data frame however all elements of a matrix are of the same type.
