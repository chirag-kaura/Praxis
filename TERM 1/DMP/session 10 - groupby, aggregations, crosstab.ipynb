```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_excel("sales.xlsx", sheet_name="Sheet2")
df
```

```python
# GroupBy
# equivalent to for each unique value the groupby field, compute the agg func on the supplied value columns
```

```python
df['sales'].groupby(df["product"]).mean()
```

```python
df['sales'].groupby([df["region"],df["product"]]).sum()
```

```python
# same as above. if there are more numeric columns, they can be included with sales.
df.groupby(['region', 'product'])[['sales', 'qty']].sum()
```

```python
# just for curiosity: multiple aggregate functions
df.groupby(['region', 'product'])['sales'].aggregate(['min', 'max'])
```

```python
# just for curiosity: this can be handy in developing reports
# groupby multiple columns, aggregate value columns using different sets of functions
df.groupby(['region', 'product']).aggregate({'sales': 'mean', 'qty': ['min', 'max']})
```

```python
for i,group in df.groupby(["product"]):
    #print(i)
    print(group)
```

```python
for i,j in df.groupby(["region","product"]):
    print(i)
    print(j)
            
```

```python
# Data aggregations
# these functions produce scalar values summarizing arrays or series objects
```

```python
df[['sales','qty','region']].groupby("region").sum()
```

```python
# .agg can be used to then feed in methods or user defined functions
df.groupby("region").agg(sum)
```

```python
df.groupby("region")['sales'].agg(["mean"])
```

```python
# define a function that summarizes grouped value column
# here it is range
def sales_summary(sales):
    return sales.max() - sales.min()
```

```python
df.groupby("region")['sales'].agg(["mean", sales_summary])
```

```python
# summary statistics
df['sales'].groupby(df["region"]).describe()
```

```python
# index speifies the row, columns the columns, values is aggregated, aggfunc is the function used to summarize values
df.pivot_table(index=["region"],columns=["product"], values=["sales"], aggfunc=["count"])
```

# crosstab
is like pivot table, it computes group frequencies

```python
# margins is to show row and col totals
pd.crosstab(df["region"], df["product"], margins=True)
```

```python
titanic = pd.read_csv("titanic.csv")
titanic.head()
```

```python
pd.crosstab(index=titanic.Sex, columns=titanic.Embarked)
```

```python
# can be used to check if survival was dependent on Sex by using Chi Square test
pd.crosstab(index=titanic.Sex, columns=titanic.Survived, margins=True)
```

```python
# similar approach as earlier. index (rows), columns (cols), values & aggfunc to populate the cross-section of rows x cols
pd.crosstab(index=titanic.Sex, columns=titanic.Survived, values=titanic.Age, aggfunc="mean", margins=True)
```
